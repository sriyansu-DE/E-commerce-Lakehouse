```sql
%python
from pyspark.sql.functions import current_timestamp, col

def write_bronze(df, table_name):
    (
        df
        .withColumn("_ingestion_timestamp", current_timestamp())
        .withColumn("_source_file", col("_metadata.file_name"))
        .write
        .format("delta")
        .mode("overwrite")
        .option("overwriteSchema", "true")
        .saveAsTable(table_name)
    )

write_bronze(customers_df, "ecommerce_lakehouse.bronze.customers")

write_bronze(orders_df, "ecommerce_lakehouse.bronze.orders")

write_bronze(order_items_df, "ecommerce_lakehouse.bronze.order_items")

write_bronze(payments_df, "ecommerce_lakehouse.bronze.payments")

write_bronze(products_df, "ecommerce_lakehouse.bronze.products")

write_bronze(sellers_df, "ecommerce_lakehouse.bronze.sellers")

write_bronze(reviews_df, "ecommerce_lakehouse.bronze.reviews")

write_bronze(geolocations_df, "ecommerce_lakehouse.bronze.geolocations")

write_bronze(category_translation_df, "ecommerce_lakehouse.bronze.category_translation")
```
