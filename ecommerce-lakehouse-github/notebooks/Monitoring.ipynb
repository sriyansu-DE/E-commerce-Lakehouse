MONITORING


```sql
CREATE TABLE IF NOT EXISTS ecommerce_lakehouse.audit.pipeline_runs
(
    pipeline_name STRING,
    layer STRING,
    status STRING,
    records_processed BIGINT,
    start_time TIMESTAMP,
    end_time TIMESTAMP,
    error_message STRING
)
USING DELTA;
```

```sql
%python
from datetime import datetime

start_time = datetime.now()

records = spark.table(
    "ecommerce_lakehouse.gold.fact_order_items"
).count()

end_time = datetime.now()

from pyspark.sql.types import StructType, StructField, StringType, LongType, TimestampType

monitor_df = spark.createDataFrame(
    [
        (
            "ecommerce_lakehouse_pipeline",
            "gold",
            "SUCCESS",
            records,
            start_time,
            end_time,
            None
        )
    ],
    StructType([
        StructField("pipeline_name", StringType(), True),
        StructField("layer", StringType(), True),
        StructField("status", StringType(), True),
        StructField("records_processed", LongType(), True),
        StructField("start_time", TimestampType(), True),
        StructField("end_time", TimestampType(), True),
        StructField("error_message", StringType(), True)
    ])
)

(
    monitor_df
    .write
    .format("delta")
    .mode("append")
    .saveAsTable(
        "ecommerce_lakehouse.audit.pipeline_runs"
    )
)
```

```sql
SELECT *

FROM ecommerce_lakehouse.audit.pipeline_runs

ORDER BY start_time DESC;
```
