```sql
SELECT
    ROUND(SUM(total_revenue), 2) AS total_revenue
FROM ecommerce_lakehouse.gold.daily_sales;
```

```sql
SELECT
    SUM(total_orders) AS total_orders
FROM ecommerce_lakehouse.gold.daily_sales;
```

```sql
SELECT
    ROUND(
        SUM(total_revenue) /
        SUM(total_orders),
        2
    ) AS average_order_value
FROM ecommerce_lakehouse.gold.daily_sales;
```

```sql
SELECT
    ROUND(
        AVG(average_review_score),
        2
    ) AS average_review_score
FROM ecommerce_lakehouse.gold.fact_order_items
WHERE average_review_score IS NOT NULL;
```

```sql
SELECT

    sales_date,

    ROUND(total_revenue, 2) AS revenue

FROM ecommerce_lakehouse.gold.daily_sales

ORDER BY sales_date;
```

```sql
SELECT

    product_category_name_english,

    ROUND(
        SUM(total_item_value),
        2
    ) AS revenue

FROM ecommerce_lakehouse.gold.fact_order_items

WHERE product_category_name_english IS NOT NULL

GROUP BY product_category_name_english

ORDER BY revenue DESC

LIMIT 10;
```

```sql
SELECT

    payment_type,

    total_orders,

    ROUND(
        total_payment_value,
        2
    ) AS payment_value

FROM ecommerce_lakehouse.gold.payment_analysis

ORDER BY payment_value DESC;
```

```sql
SELECT

    customer_state,

    ROUND(
        SUM(total_item_value),
        2
    ) AS revenue

FROM ecommerce_lakehouse.gold.fact_order_items

GROUP BY customer_state

ORDER BY revenue DESC;
```

```sql
SELECT

    average_review_score,

    COUNT(*) AS records

FROM ecommerce_lakehouse.gold.fact_order_items

WHERE average_review_score IS NOT NULL

GROUP BY average_review_score

ORDER BY average_review_score;
```

```sql
SELECT

    delivery_status,

    COUNT(*) AS total_orders

FROM ecommerce_lakehouse.gold.delivery_performance

GROUP BY delivery_status;
```

```sql
SELECT

    seller_id,

    seller_state,

    ROUND(total_revenue, 2) AS revenue,

    total_orders,

    ROUND(average_review_score, 2)
        AS average_review_score

FROM ecommerce_lakehouse.gold.seller_performance

ORDER BY revenue DESC

LIMIT 20;
```

```sql
SELECT

    CASE

        WHEN total_spend >= 5000
            THEN 'HIGH_VALUE'

        WHEN total_spend >= 1000
            THEN 'MEDIUM_VALUE'

        ELSE 'LOW_VALUE'

    END AS customer_segment,

    COUNT(*) AS customers,

    ROUND(
        AVG(total_spend),
        2
    ) AS average_spend

FROM ecommerce_lakehouse.gold.customer_360

GROUP BY

    CASE

        WHEN total_spend >= 5000
            THEN 'HIGH_VALUE'

        WHEN total_spend >= 1000
            THEN 'MEDIUM_VALUE'

        ELSE 'LOW_VALUE'

    END;
```
