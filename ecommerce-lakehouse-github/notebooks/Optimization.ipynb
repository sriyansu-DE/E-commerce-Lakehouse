```sql
CREATE OR REPLACE TABLE ecommerce_lakehouse.gold.fact_order_items_partitioned
USING DELTA
PARTITIONED BY (purchase_date)
AS
SELECT
    *,
    DATE(order_purchase_timestamp) AS purchase_date
FROM ecommerce_lakehouse.gold.fact_order_items;

OPTIMIZE ecommerce_lakehouse.gold.fact_order_items_partitioned;
```
